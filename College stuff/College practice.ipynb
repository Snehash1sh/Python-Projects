```python
class Employee:
    def __init__(self, emp_id, emp_name):
        self.emp_id = emp_id
        self.emp_name = emp_name


class Salary:
    def __init__(self, basic_salary):
        self.basic_salary = basic_salary

    def calculate_salary(self):
        self.hra = self.basic_salary * 0.20
        self.da = self.basic_salary * 0.10
        self.total_salary = self.basic_salary + self.hra + self.da


class EmployeeDetails(Employee, Salary):
    def __init__(self, emp_id, emp_name, basic_salary):
        Employee.__init__(self, emp_id, emp_name)
        super(Employee, self).__init__(basic_salary)

    def display(self):
        self.calculate_salary()

        print("Employee ID:", self.emp_id)
        print("Employee Name:", self.emp_name)
        print("Basic Salary:", self.basic_salary)
        print("HRA (20%):", self.hra)
        print("DA (10%):", self.da)
        print("Total Salary:", self.total_salary)


emp = EmployeeDetails(101, "Rahul", 30000)

emp.display()
```

```python
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display(self):
        print("Student Name:", self.name)
        print("Roll Number:", self.roll_no)


class Science(Student):
    def __init__(self, name, roll_no, physics, chemistry):
        super().__init__(name, roll_no)
        self.physics = physics
        self.chemistry = chemistry

    def display(self):
        super().display()
        print("Physics Marks:", self.physics)
        print("Chemistry Marks:", self.chemistry)
        print("Total Marks:", self.physics + self.chemistry)


class Commerce(Student):
    def __init__(self, name, roll_no, accountancy, economics):
        super().__init__(name, roll_no)
        self.accountancy = accountancy
        self.economics = economics

    def display(self):
        super().display()
        print("Accountancy Marks:", self.accountancy)
        print("Economics Marks:", self.economics)
        print("Total Marks:", self.accountancy + self.economics)


# Creating Science student
s1 = Science("Rahul", 101, 85, 90)

# Creating Commerce student
s2 = Commerce("Priya", 102, 88, 92)

print("---- Science Student Details ----")
s1.display()

print("\n---- Commerce Student Details ----")
s2.display()
```
